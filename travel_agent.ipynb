This agent recommends the best travel destination based on the user's **budget**, **available days**, and **preferred trip type** (Beach, Mountain, City, Adventure). It follows the classic **Sense → Think → Act** agent cycle.

## Step 1: Define the destinations database (Environment)

```python
destinations = [
    {"place": "Murree",       "cost": 15000, "days": 2, "type": "Mountain"},
    {"place": "Hunza Valley", "cost": 45000, "days": 5, "type": "Mountain"},
    {"place": "Karachi Beach","cost": 12000, "days": 2, "type": "Beach"},
    {"place": "Gwadar",       "cost": 30000, "days": 4, "type": "Beach"},
    {"place": "Lahore City",  "cost": 10000, "days": 2, "type": "City"},
    {"place": "Islamabad",    "cost": 18000, "days": 3, "type": "City"},
    {"place": "Skardu",       "cost": 50000, "days": 6, "type": "Adventure"},
    {"place": "Neelum Valley","cost": 35000, "days": 4, "type": "Adventure"},
]
```

## Step 2: Filter destinations that match the user's needs (Perception + Reasoning)

```python
def filter_destinations(dest_list, budget, available_days, trip_type):
    suitable = []
    for dest in dest_list:
        if (dest["cost"] <= budget and
                dest["days"] <= available_days and
                dest["type"] == trip_type):
            suitable.append(dest)
    return suitable
```

## Step 3: Choose the best destination among suitable ones (Decision Making)

```python
def recommend_best_destination(dest_list):
    if not dest_list:
        return None
    best_dest = dest_list[0]
    for dest in dest_list:
        if dest["cost"] > best_dest["cost"]:
            best_dest = dest
    return best_dest
```

## Step 4: The Agent function itself (Action)

```python
def travel_agent(budget, available_days, trip_type):
    print("Searching best travel destination for your trip...")
    suitable_destinations = filter_destinations(destinations, budget, available_days, trip_type)
    best_destination = recommend_best_destination(suitable_destinations)

    if best_destination:
        print(f"Recommended Destination: {best_destination['place']}")
        print(f"Trip Type: {best_destination['type']}")
        print(f"Duration: {best_destination['days']} days")
        print(f"Estimated Cost: PKR {best_destination['cost']}")
    else:
        print("No suitable destination found within your budget, days, and preference.")
```

## Step 5: Run the agent with a sample user request

```python
travel_agent(budget=40000, available_days=5, trip_type="Mountain")
```
